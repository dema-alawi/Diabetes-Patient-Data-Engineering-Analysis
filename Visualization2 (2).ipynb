```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
df = pd.read_csv('/diabetic_data.csv')
```

```python
plt.figure(figsize=(7,5))

counts = df['readmitted'].value_counts()

plt.bar(counts.index, counts.values)

plt.title('Distribution of Readmission')

plt.xlabel('Readmission Status')

plt.ylabel('Number of Patients')

plt.show()
```

```python
#هون bar chart بدنا نشوف علاقه الجنس بالعوده للمستشفى؟
gender_read = pd.crosstab(df['gender'], df['readmitted'])

gender_read.plot(kind='bar', figsize=(8,5))

plt.title('Gender vs Readmission')

plt.xlabel('Gender')

plt.ylabel('Count')

plt.show()
```

```python
#مقاؤنه بين اذا المريض حيرجع ع المستشفى بناءا ع عدد الادويه الي باخدها؟
sns.boxplot(x='readmitted', y='num_medications', data=df)
```

```python
#هون بدنا نشوف اذا الايام الي قضاهم بالمستشفى اول مره بأثر على انو رح يرجع عالستشفى او لا؟
sns.boxplot(x='readmitted', y='time_in_hospital', data=df)
```

```python
#هون heat map عشان نعرف علاقات الاعمده المؤثرة على العوده الى المستشفى مع بعض؟
plt.figure(figsize=(12,8))

numeric_df = df.select_dtypes(include=['int64','float64'])

sns.heatmap(numeric_df.corr(), cmap='coolwarm')

plt.show()
```
