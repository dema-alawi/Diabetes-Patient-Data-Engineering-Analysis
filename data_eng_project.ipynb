```python
# استدعاء المكتبات
# هون بنستدعي المكتبات اللي رح نستخدمها بالمشروع

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# تحميل ملف البيانات
df = pd.read_csv("/diabetic_data.csv")
# عرض أول 5 صفوف عشان نتأكد انو الداتا انقرأت صح

print("First 5 Rows")
display(df.head())

# عرض أسماء الأعمدة

print("Column Names")

print(df.columns)

# تحويل ? لقيم فارغة
# بعض القيم المفقودة كانت مكتوبة ؟
# هون حولناها لـ NaN

df.replace('?', np.nan, inplace=True)

# حذف الأعمدة غير المهمة
# حذف أعمدة ما إلها فايدة بالتحليل

df.drop(
    columns=['weight', 'payer_code'],
    inplace=True,
    errors='ignore'
)


# تنظيف الأعمدة الرقمية
# هون بنحول الأعمدة الرقمية لأرقام فعلية
# وإذا في قيم غلط بنحولها لـ NaN
# وبعدها نعبيها بـ 0

numeric_cols = [

    'number_outpatient',
    'number_emergency',
    'number_inpatient',
    'num_medications',
    'number_diagnoses'

]

for col in numeric_cols:

    df[col] = pd.to_numeric(

        df[col],
        errors='coerce'

    ).fillna(0).astype(int)

    print(f"Cleaned Column: {col}")


# معالجة القيم الفارغة بالنصوص
# هون إذا في قيم ناقصة بالأعمدة النصية
# بنعبيها بأكثر قيمة متكررة

categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:

    df[col].fillna(
        df[col].mode()[0],
        inplace=True
    )

# تحويل أعمدة التشخيص لنصوص
# تشخيصات المرض لازم تكون string
# عشان نقدر نعمل عليها تصنيف

for col in ['diag_1', 'diag_2', 'diag_3']:

    df[col] = df[col].astype(str)

# تحويل العمر لأرقام
# العمر بالداتا عبارة عن ranges
# هون حولناه لأرقام midpoint

age_map = {

    '[0-10)': 5,
    '[10-20)': 15,
    '[20-30)': 25,
    '[30-40)': 35,
    '[40-50)': 45,
    '[50-60)': 55,
    '[60-70)': 65,
    '[70-80)': 75,
    '[80-90)': 85,
    '[90-100)': 95

}

df['age_midpoint'] = df['age'].map(age_map)

print("Age Transformation Completed")

display(df[['age', 'age_midpoint']].head())

# إنشاء مجموع الزيارات
# هون جمعنا كل أنواع الزيارات
# عشان نعرف قديش المريض تعامل مع المستشفى

df['total_visits'] = (

    df['number_outpatient'] +
    df['number_emergency'] +
    df['number_inpatient']

)

print("Total Visits Feature Created")

display(

    df[[
        'number_outpatient',
        'number_emergency',
        'number_inpatient',
        'total_visits'
    ]].head()

)

# تصنيف التشخيصات
# هون بنحول أكواد التشخيص
# لفئات طبية مفهومة

def categorize_diagnosis(code):

    if code.startswith('250'):

        return 'Diabetes'

    elif code.startswith('V') or code.startswith('E'):

        return 'Supplementary'

    else:

        try:

            code = float(code)

            if 390 <= code < 460:

                return 'Circulatory'

            elif 460 <= code < 520:

                return 'Respiratory'

            elif 520 <= code < 580:

                return 'Digestive'

            elif 800 <= code < 1000:

                return 'Injury'

            else:

                return 'Other'

        except:

            return 'Unknown'


# تطبيق التصنيف
# هون طبقنا التصنيف على كل أعمدة التشخيص

for col in ['diag_1', 'diag_2', 'diag_3']:

    df[col + '_cat'] = df[col].apply(
        categorize_diagnosis
    )

print("Diagnosis Categorization Completed")

display(

    df[[
        'diag_1',
        'diag_1_cat',
        'diag_2',
        'diag_2_cat',
        'diag_3',
        'diag_3_cat'
    ]].head()

)

# هل المريض عنده سكري؟
# إذا أي تشخيص كان Diabetes
# نعطي القيمة 1

df['has_diabetes'] = (

    (df['diag_1_cat'] == 'Diabetes') |
    (df['diag_2_cat'] == 'Diabetes') |
    (df['diag_3_cat'] == 'Diabetes')

).astype(int)

print("Diabetes Feature Created")
# حساب عدد الأمراض المزمنة
# هون حسبنا عدد مرات ظهور السكري
# بالتشخيصات الثلاثة

df['chronic_category_count'] = (

    (df['diag_1_cat'] == 'Diabetes').astype(int) +
    (df['diag_2_cat'] == 'Diabetes').astype(int) +
    (df['diag_3_cat'] == 'Diabetes').astype(int)

)

print("Chronic Category Count Created")

display(

    df[[
        'diag_1_cat',
        'diag_2_cat',
        'diag_3_cat',
        'chronic_category_count'
    ]].head()

)

# تحديد المرضى عاليي الخطورة
# إذا عدد الأمراض المزمنة 3 أو أكثر
# يعتبر High Risk

df['high_risk'] = (

    df['chronic_category_count'] >= 3

).astype(int)

print("High Risk Feature Created")

display(

    df[[
        'chronic_category_count',
        'high_risk'
    ]].head()

)

# تحويل عدد التشخيصات لمستويات
# هون حولنا عدد التشخيصات
# لمستويات خطورة

def diag_level(x):

    if x <= 3:

        return "Low"

    elif x <= 6:

        return "Medium"

    else:

        return "High"

df['diagnosis_level'] = (

    df['number_diagnoses']
    .apply(diag_level)

)

print("Diagnosis Level Created")

display(

    df[[
        'number_diagnoses',
        'diagnosis_level'
    ]].head()

)

# One Hot Encoding
# هون حولنا الفئات النصية لأعمدة رقمية

df = pd.get_dummies(

    df,

    columns=[
        'diag_1_cat',
        'diag_2_cat',
        'diag_3_cat'
    ],

    prefix=[
        'diag1',
        'diag2',
        'diag3'
    ]

)

print("One Hot Encoding Applied")

# تحديد إذا الدخول طارئ
# إذا نوع الدخول 1 أو 2
# يعتبر Emergency

def categorize_admission(code):

    if code in [1, 2]:

        return 1

    else:

        return 0

df['is_emergency'] = (

    df['admission_type_id']
    .apply(categorize_admission)

)

print("Emergency Feature Created")

display(

    df[[
        'admission_type_id',
        'is_emergency'
    ]].head()

)

# تحديد مستوى الخطورة
# هون عملنا تصنيف خطورة
# حسب عدد التشخيصات

def get_risk(count):

    if count <= 3:

        return 'Low'

    elif count <= 6:

        return 'Medium'

    else:

        return 'High'

df['risk_level'] = (

    df['number_diagnoses']
    .apply(get_risk)

)

print("Risk Level Feature Created")

display(

    df[[
        'number_diagnoses',
        'risk_level'
    ]].head()

)

# تبسيط التخصصات الطبية
# هون دمجنا تخصصات كثيرة
# بفئات أبسط وأسهل للتحليل

def simplify_specialty(value):

    value = str(value).lower()

    if value == "?":

        return "Missing"

    elif "surgery" in value:

        return "Surgery"

    elif 'internalmedicine' in value or 'cardiology' in value:

        return 'Internal_Med_Heart'

    else:

        return "Other"

df['specialty_new'] = (

    df['medical_specialty']
    .apply(simplify_specialty)

)

print("Medical Specialty Simplified")

display(

    df[[
        'medical_specialty',
        'specialty_new'
    ]].head(20)

)

# ==========================================
# تحويل القيم الثنائية لأرقام
# ==========================================

# هون حولنا القيم النصية
# لـ 0 و 1

def convert_to_binary(status):

    if status in ['Ch', 'Yes']:

        return 1

    else:

        return 0

df['change_numeric'] = (

    df['change']
    .apply(convert_to_binary)

)

df['diabetesMed_numeric'] = (

    df['diabetesMed']
    .apply(convert_to_binary)

)

print("Binary Conversion Completed")

display(

    df[[
        'change',
        'change_numeric',
        'diabetesMed',
        'diabetesMed_numeric'
    ]].head(10)

)

# ==========================================
# تحديد عدم استقرار الإنسولين
# ==========================================

# إذا الإنسولين Up أو Down
# يعني في عدم استقرار

def encode_insulin(value):

    if value in ["Up", "Down"]:

        return 1

    else:

        return 0

df['insulin_instability'] = (

    df['insulin']
    .apply(encode_insulin)

)

print("Insulin Instability Feature Created")

display(

    df[[
        'insulin',
        'insulin_instability'
    ]].head()

)

# ==========================================
# تحويل readmitted لأرقام
# ==========================================

# إذا المريض رجع للمستشفى
# نعطي القيمة 1

df['readmitted_binary'] = (

    df['readmitted']
    .apply(lambda x: 0 if x == 'NO' else 1)

)

print("Readmitted Column Converted")

display(

    df[[
        'readmitted',
        'readmitted_binary'
    ]].head()

)

# ==========================================
# وصف إحصائي
# ==========================================

# عرض إحصائيات عامة للداتا

print("Statistical Summary")

display(df.describe())

# ==========================================
# فحص القيم الفارغة
# ==========================================

# التأكد إذا ضل قيم ناقصة

print("Missing Values Check")

print(df.isnull().sum())

# ==========================================
# رسم بياني لمستويات الخطورة
# ==========================================

# رسم توزيع مستويات الخطورة

df['risk_level'].value_counts().plot(
    kind='bar'
)

plt.title("Risk Level Distribution")

plt.xlabel("Risk Level")

plt.ylabel("Count")

plt.show()

# ==========================================
# حفظ الملف النهائي
# ==========================================

# حفظ الداتا بعد التنظيف والتحويل

df.to_csv(

    'processed_diabetic_data.csv',
    index=False

)

print("Dataset Saved Successfully")
```
